```python
def calc_pairwise_scores(rankings, num_entities):
    pairwise_scores = np.zeros((num_entities, num_entities))
    for person in rankings:
        to_check = [i for i in range(num_entities)]
        for element in person:
            to_check.remove(element)
            if len(to_check) > 0:
                pairwise_scores[element, to_check] += 1
    return pairwise_scores
```

#### Borda Count

```python
def borda_count(rankings, num_entities):
    borda = [i for i in reversed(range(num_entities))]

    counts = {i:0 for i in borda}

    for row in rankings:
        for c in range(num_entities):
            counts[row[c]] += borda[c]

    result = sorted(counts.items(), reverse=True, key=lambda item: item[1])
    results = [i[0] for i in result]
    return results
```

#### Copeland's method

```python
def copeland_method(rankings, num_entities):
    pairwise_scores = calc_pairwise_scores(rankings=rankings, num_entities=num_entities)

    copeland = np.zeros((num_entities, num_entities))
    for i in range(len(pairwise_scores)):
        for j in range(len(pairwise_scores[i])):
            if pairwise_scores[i, j] > pairwise_scores[j, i]:
                copeland[i, j] = 1
            elif pairwise_scores[i, j] == pairwise_scores[j, i]:
                copeland[i, j] = 0.5

    counts_copeland = np.sum(copeland, axis=1)
    sorted_list = sorted([(c, counts_copeland[c]) for c in range(len(counts_copeland))], reverse=True, key=lambda item: item[1])
    results = [i[0] for i in sorted_list]
    return results
```

```python
borda_ranks = borda_count(rankings=rankings, num_entities=14)
copeland_ranks = copeland_method(rankings=rankings, num_entities=14)

print(" ------------------------------------- ")

print(" Ranked List by Borda Counts:")
for c in range(14):
    print(f" {c}: {projects[borda_ranks[c]]}")

print(" ------------------------------------- ")

print(" Ranked List by Copeland Ranks:")
for c in range(14):
    print(f" {c}: {projects[copeland_ranks[c]]}")

print(" ------------------------------------- ")
```
